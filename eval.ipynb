```python
import os
import pandas as pd
import numpy as np

path = os.path.join("D:/Dani/relbench/relbench/burger_data/", "burger")
df = os.path.join(path, "test2.csv")
df1 = pd.read_csv(df)
```

```python
df1
```

```python
# Extract unique customers_id
unique_customers = df1['customers_id'].unique()

# Create a new DataFrame with the unique customers_id and assign label 0
df1 = pd.DataFrame({'customers_id': unique_customers, 'label': 0})

# Display the resulting DataFrame
print(df1.head())
```

```python
path2 = os.path.join("C:/Users/KN2C/Desktop/Dani/", "relbench")
dff = os.path.join(path2, "result1.csv")
df2 = pd.read_csv(dff)
```

```python
df2
```

```python
df2 = df2[['mapped_customers_id', 'label']]

df2.rename(columns={'mapped_customers_id': 'customers_id'}, inplace=True)

# Keep only 'customers_id' and 'label' columns
df2 = df2[['customers_id', 'label']]

# Display the updated DataFrame
print(df2)
```

```python
merged_df = pd.merge(df1, df2, on='customers_id', suffixes=('_df1', '_df2'))

# Compare labels and calculate accuracy
merged_df['correct'] = merged_df['label_df1'] == merged_df['label_df2']
accuracy = merged_df['correct'].mean()

# Display results
print(f"Accuracy: {accuracy * 100:.2f}%")
print("\nMerged Data with Comparison:")
print(merged_df)
```

```python
# Filter df1 to include only customers_id that are present in df2
filtered_df1 = df1[df1['customers_id'].isin(df2['customers_id'])]

# Merge the two DataFrames on 'customers_id'
merged_df = pd.merge(filtered_df1, df2, on='customers_id', how='outer', suffixes=('_df1', '_df2'))

# Assign label 1 for customers in df2 but not in df1
merged_df['label_df1'] = merged_df['label_df1'].fillna(1)

# Compare labels and calculate accuracy
merged_df['correct'] = merged_df['label_df1'] == merged_df['label_df2']
accuracy = merged_df['correct'].mean()

# Display results
print(f"Accuracy: {accuracy * 100:.2f}%")
print("\nMerged Data with Comparison:")
print(merged_df)
```

```python
from sklearn.metrics import f1_score

# Filter df1 to include only customers_id that are present in df2
filtered_df1 = df1[df1['customers_id'].isin(df2['customers_id'])]

# Merge the two DataFrames on 'customers_id'
merged_df = pd.merge(filtered_df1, df2, on='customers_id', how='outer', suffixes=('_df1', '_df2'))

# Assign label 1 for customers in df2 but not in df1
merged_df['label_df1'] = merged_df['label_df1'].fillna(1)

# Extract true labels and predicted labels
true_labels = merged_df['label_df2']
predicted_labels = merged_df['label_df1']

# Calculate F1 score
f1 = f1_score(true_labels, predicted_labels)

# Display F1 score as a percentage
print(f"F1 Score: {f1 * 100:.2f}%")
```
